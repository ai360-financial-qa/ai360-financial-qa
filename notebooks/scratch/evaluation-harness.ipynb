```python
from financial_qa.base import BaseAgentLoop

from financial_qa.chunkers import SlidingWindowChunker
from financial_qa.embedders import HFEmbedder
from financial_qa.rag import RAG
from agent_loop import OpenRouterAgentLoop

from financial_qa.evaluation import evaluate, load_jsonl

import os
import dotenv
```

```python
dotenv.load_dotenv('.env')
OPENROUTER_API_KEY = os.getenv('OPENROUTER_API_KEY')
```

```python
dataset = load_jsonl('dataset.jsonl')
```

```python
from tqdm import tqdm

def evaluate_loop(loop: BaseAgentLoop) -> None:
    huyhuyhuy = {}
    for k, v in list(dataset.items())[:]:
        huyhuyhuy[k] = v
    predicted = {k: {"question_id": q["question_id"], "question": q["question"], "answer": loop.query(q)[0]} for k, q in tqdm(huyhuyhuy.items())}
    result = evaluate(golden=huyhuyhuy, predicted=predicted, api_key=OPENROUTER_API_KEY, include_evidence=False)
    print(f'Score = {result} correct out of {len(dataset)} = {result / len(huyhuyhuy)}% score')
```

```python
print(dataset)
```

```python
fixed_size_chunker = SlidingWindowChunker()
hf_embedder = HFEmbedder(model_name="intfloat/multilingual-e5-small")
```

```python
openrouter_agent_loop = OpenRouterAgentLoop(rag=RAG(fixed_size_chunker, hf_embedder))
```

```python
evaluate_loop(openrouter_agent_loop)
```
